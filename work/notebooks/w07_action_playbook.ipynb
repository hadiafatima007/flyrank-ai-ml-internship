## 1. Ranked actions + reason codes

The ranked queue is a decision-support list for human review. Items are ordered by the baseline action score, with the reason code explaining why an item received its recommended action.

### Reason codes and actions

| Reason code | Action | Why it is ranked |
|---|---|---|
| STALE_REFRESH | Refresh Content | The content has been inactive for a long period, so a human should check whether its information, structure, or search intent needs updating. |
| QUICK_WIN | Optimise Existing Content | The content shows an existing performance opportunity that may be addressable without a full rewrite. |

### Archetype → action mapping

- **Stale content → STALE_REFRESH → Refresh Content**
- **Existing content with a potential quick opportunity → QUICK_WIN → Optimise Existing Content**
- **Ambiguous or weak signals → Human review before assigning an action**

The score is a prioritisation signal, not a probability of success or a guaranteed estimate of business value.

The reason code explains the signal behind the recommendation. It does not prove that the recommended action will improve performance.

```python
from pathlib import Path
import pandas as pd

queue_path = Path("../outputs/baseline_action_score.csv")

queue = pd.read_csv(queue_path)

print("Ranked queue loaded successfully.")
print("Rows:", len(queue))
print("Columns:", list(queue.columns))

display(queue.head(20))
```

### Ranked action queue

The ranked queue prioritizes content using the baseline action score. Higher-scoring items are reviewed first.

The main observed action in the top 20 is **Refresh Content**, with the reason code **STALE_REFRESH**. A smaller number of items are classified as **Optimise Existing Content** with **QUICK_WIN**.

The reason codes provide a simple explanation for why an item appears in the queue. They are intended as decision-support signals rather than automatic instructions.

The queue should be reviewed by a human before any content change is made, especially when search demand is low or missing, trend data is missing, or the observed decline may have another explanation.

## 2. Intended use and limits

This playbook is intended to help a content team prioritize which pages to review first and what type of action may be worth considering.

The scores and reason codes are decision-support signals. They do not prove that refreshing or optimising a page will improve performance.

The recommendations are most useful when the underlying search, ranking, traffic, and content signals are available and reasonably current.

The output should not be treated as an automatic publishing or deletion system. A human should review the content and its context before taking action.

```python
# Basic summary of the decision-support queue

print("Number of content items:", len(queue))

print("\nAction distribution:")
display(
    queue["action_label"]
    .value_counts()
    .rename_axis("action")
    .reset_index(name="count")
)

print("\nReason-code distribution:")
display(
    queue["reason_code"]
    .value_counts()
    .rename_axis("reason_code")
    .reset_index(name="count")
)

print("\nScore summary:")
print(queue["score"].describe())
```

## 3. Human review + the no-go list

### Human review rules

Before taking action on a recommendation, a human reviewer should:

1. Check that the content is still relevant to the target search intent.
2. Review the current search demand and ranking signals.
3. Check whether the observed decline could be seasonal or temporary.
4. Check for missing or unreliable input data.
5. Confirm that the recommended action makes sense for the content type.
6. Review the actual page before making any content changes.

The score and reason code should be treated as decision-support signals, not as automatic decisions.

### What should NOT be automated

The system should not automatically:

- Publish content changes.
- Delete or unpublish content.
- Change a page's search intent.
- Rewrite an entire article without human review.
- Make decisions based only on a single score.
- Treat missing data as evidence that content is performing poorly.
- Assume that every traffic decline means the content needs a refresh.

A human should remain responsible for the final action.

```python
# Human-review rules and no-go cases

review_rules = {
    "priority": "Review higher-scoring items first.",
    "reason_check": "Confirm that the reason code matches the current content situation.",
    "data_check": "Check for missing, unusual, or stale inputs.",
    "content_check": "Review the actual page before taking action.",
    "external_factors": "Check for seasonality, campaigns, or other external changes.",
    "final_decision": "A human reviewer makes the final decision."
}

no_go_cases = [
    "Automatic publishing or editing",
    "Automatic deletion of content",
    "Automatically changing search intent",
    "Claiming guaranteed business impact",
    "Overriding human reviewers",
    "Acting when important inputs are missing or stale"
]

print("Human review rules:")
for key, value in review_rules.items():
    print(f"- {key}: {value}")

print("\nNo-go cases:")
for item in no_go_cases:
    print(f"- {item}")
```

## 4. Monitoring / retrain triggers

The recommendations should be monitored because search behavior, rankings, and content performance can change over time.

Useful monitoring signals include:

- A noticeable change in the distribution of action labels or reason codes.
- A change in the score distribution compared with the data used to build the baseline.
- Increasing amounts of missing input data.
- A measured drop in model performance on newly available labelled data.
- Changes in search demand or ranking patterns that make the original rules less representative.

A retraining or review should be considered when these changes are persistent rather than caused by a short-term fluctuation.

The model should also be reviewed if human reviewers repeatedly disagree with the recommended actions. This can indicate that the features or decision rules are no longer aligned with practical content decisions.

These are monitoring and decision-support triggers, not automatic retraining rules.

```python
# Simple monitoring summary for the current queue

print("Current queue size:", len(queue))

print("\nScore distribution:")
print(queue["score"].describe())

print("\nReason-code distribution:")
display(
    queue["reason_code"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
    .rename("percentage")
    .to_frame()
)

print("\nMissing values in important monitoring fields:")
monitor_cols = [
    "score",
    "reason_code",
    "action_label"
]

available_monitor_cols = [
    col for col in monitor_cols if col in queue.columns
]

display(queue[available_monitor_cols].isna().sum())
```

## 5. Exports for the paper

The ranked queue is exported to `work/outputs/` so that the paper can reuse the same decision-support output.

The exported queue contains the content identifier, action, reason code, and score needed to reproduce the prioritisation view.

The queue is an output for analysis and human review. It is not a production automation file.

```python
from pathlib import Path

# Create the output directory if it does not already exist
output_dir = Path("../outputs")
output_dir.mkdir(parents=True, exist_ok=True)

# Select the fields needed for the ranked decision-support queue
export_cols = [
    "content_id",
    "score",
    "reason_code",
    "action_label"
]

export_cols = [col for col in export_cols if col in queue.columns]

ranked_queue = queue[export_cols].copy()

# Rank from highest to lowest score
ranked_queue = ranked_queue.sort_values(
    "score",
    ascending=False
).reset_index(drop=True)

# Export queue
output_path = output_dir / "action_playbook_queue.csv"

ranked_queue.to_csv(output_path, index=False)

print("Exported:", output_path)
print("Rows exported:", len(ranked_queue))

display(ranked_queue.head(20))
```
