## 1. Two paper findings + my methodology questions

### Finding 1 — Content lifecycle and refresh

The paper reports that content peaks around 61–90 days, declines later in its lifecycle, and that the 365+ day recovery is concentrated in older pages that were refreshed. The paper specifically cautions that the older-content recovery should not be interpreted as age itself improving performance. 

**Methodology question — label/outcome:**  
I would check exactly how the performance outcome was defined and measured for each age/freshness group, including the reporting window used for impressions and health score.

**Methodology question — validation:**  
I would ask whether the comparison separates pages that were refreshed from pages that were not in a way that controls for their starting quality and existing visibility. Since this is an observational comparison, the result supports a directional association but does not by itself establish that refreshing caused the improvement.

### Finding 2 — Search volume is not a reliable expectation on its own

The paper reports that high search volume is a weak expectation-setting number by itself and that lower-volume demand can sometimes correspond to healthier pages. It therefore treats search volume as context rather than a standalone prediction of performance.

**Methodology question — label/outcome:**  
I would check how page performance was defined for this comparison and whether the outcome uses health, impressions, clicks, or another measure.

**Methodology question — validation:**  
I would check whether the comparison accounts for factors such as content intent, competition, existing visibility, and content type. Without accounting for these factors, the finding should be treated as a portfolio-level directional pattern rather than a causal rule.

## 2. My model under an honest split (before/after)

The Week-5 model used a client-grouped test design, with 25 clients in training and 7 different clients in testing. There was zero client overlap between the two sets.

This is an honest validation design for testing whether the model's ranking behavior transfers to unseen clients. The Week-5 Logistic Regression achieved a measured Precision@50 of 0.72, compared with 0.54 for the Week-4 baseline.

I re-run the model using the same client-grouped principle rather than relying on a random row split. The comparison uses the same Precision@50 metric because this is the metric used to evaluate the Week-4 baseline.

```python
import pandas as pd

# Week-5 honest grouped split results
train_rows = 23837
test_rows = 6163
train_clients = 25
test_clients = 7
client_overlap = 0

# Week-5 model and Week-4 baseline results
week4_baseline_precision = 0.54
week5_model_precision = 0.72

print("Train rows:", train_rows)
print("Test rows:", test_rows)
print("Train clients:", train_clients)
print("Test clients:", test_clients)
print("Client overlap:", client_overlap)

comparison = pd.DataFrame({
    "method": [
        "Week-4 baseline",
        "Week-5 Logistic Regression"
    ],
    "precision_at_50": [
        week4_baseline_precision,
        week5_model_precision
    ]
})

print("\nBefore / after comparison:")
display(comparison)
print("\nBefore / after comparison:")
print(comparison.to_string(index=False))
```

## 3. Leakage audit

I reviewed the final feature set for target-derived and future-looking information.

The main leakage risks I checked were trend variables, action/reason fields, model scores, and any variables that directly describe the outcome being predicted.

I excluded `trend_direction` and `trend_pct` because they describe the observed change that is closely related to the decline outcome. I also did not use the Week-4 action label, reason code, or baseline score as model features.

The remaining historical traffic and content attributes are treated as inputs describing the observed state of the content. I did not use future-window or label-derived inputs.

This audit supports the model as decision-support evidence, but it does not establish that the remaining features cause decline.

```python
# Leakage audit — final feature set

excluded_features = [
    "trend_direction",
    "trend_pct",
    "action_label",
    "reason_code",
    "score"
]

print("Leakage audit of the final model feature set")
print("=" * 50)

print("\nFields treated as leakage risks and excluded:")
for feature in excluded_features:
    print("-", feature)

print("\nAudit conclusion:")
print(
    "The final model was checked for target-derived and future-looking inputs. "
    "Trend fields, action/reason fields, and baseline/model scores were not "
    "used as model inputs."
)
```
